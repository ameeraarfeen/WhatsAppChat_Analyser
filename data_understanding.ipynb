```python
import re
import pandas as pd
```

```python
f=open('WhatsAppGroupChat.txt','r', encoding='utf-8') #utf-8 To help Python correctly read emojis and multilingual text
```

```python
data=f.read()
```

```python
print(type(data))
```

```python

pattern = r'\d{1,2}/\d{1,2}/\d{2,4},\s\d{1,2}:\d{2}\s-\s' #splitting the wtsap msg between date and msg
```

```python
messages= re.split(pattern, data)[1:] #using [1:] to remove the space at the start
```

```python
dates=re.findall(pattern, data)
```

```python
#creating pandas dataframe
df = pd.DataFrame({
    'user_message': messages,
    'date': dates
})
```

```python
#converting the date column 
df['date'] = pd.to_datetime(
    df['date'],
    format='%d/%m/%y, %H:%M - '
)
```

```python
df.head()
```

```python
#separating user and msgs
users = []
messages_only = []

for message in df['user_message']:
    entry = re.split(r'([\w\W]+?):\s', message)

    if len(entry) >= 3:
        users.append(entry[1])
        messages_only.append(entry[2])
    else:
        users.append('group_notification')
        messages_only.append(entry[0])
df['user'] = users
df['message'] = messages_only
```

```python
df.head()
```

```python
#making columns for year month and date hour min
df['year']=df['date'].dt.year
df['month']=df['date'].dt.month_name()
df['day']=df['date'].dt.day
df['hour']=df['date'].dt.hour
df['minute']=df['date'].dt.minute
```
